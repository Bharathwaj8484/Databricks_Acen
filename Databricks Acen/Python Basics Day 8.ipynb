```python
#9. sum/avg:
tot = 0
cnt = 0
while True:
    inp = input("num or done: ")
    if inp.lower() == "done":
        break
    try:
        tot += float(inp)
        cnt += 1
    except:
        continue
print(tot, tot / cnt if cnt else 0)
```

```python
#15. triangle:
s1 = float(input("s1: "))
s2 = float(input("s2: "))
s3 = float(input("s3: "))
if s1 + s2 > s3 and s1 + s3 > s2 and s2 + s3 > s1:
    if s1 == s2 == s3:
        print("equilateral")
    elif s1 == s2 or s1 == s3 or s2 == s3:
        print("isosceles")
    else:
        print("scalene")
else:
    print("invalid")
```

```python
#20. sum squares:
def sos(n):
    s = 0
    for d in str(n):
        s += int(d) ** 2
    return s
print(sos(1))
print(sos(9))
print(sos(11))
print(sos(121))
```

```python
#21. Fibonacci:
a, b = 0, 1
fib = []
while a <= 50:
    fib.append(a)
    a, b = b, a + b
fib = [x for x in fib if x != 0 and x <= 34]
print(" ".join(map(str, fib)))
```

```python
#22. primes:
for num in range(2, 101):
    prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            prime = False
            break
    if prime:
        print(num, end=" ")
print()
```

```python
#27. list sum:
lst = [10, 20, 30, 40, 50]
s = 0
i = 0
while i < len(lst):
    inp = input("Enter or break: ")
    if inp.lower() == "break":
        break
    s += lst[i]
    i += 1
print(s)
```

```python
#28. reverse:
num = int(input("num: "))
rev = 0
temp = abs(num)
while temp:
    rev = rev * 10 + temp % 10
    temp //= 10
if num < 0:
    rev = -rev
print(rev)
```
