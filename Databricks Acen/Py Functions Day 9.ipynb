```python
#1. max in list:
def find_max(lst):
    return max(lst)
print(find_max([4,55,66,77,88,89,444]))
```

```python
#2. nested factorial:
def nested_fact(n):
    def fact(x):
        res = 1
        for i in range(1, x+1):
            res *= i
        return res
    return fact(n)
print(nested_fact(5))
```

```python
#3. lambda min:
min_lambda = lambda a, b, c: min(a, b, c)
print(min_lambda(3, 1, 2))
```

```python
#4. reverse string & alternate chars:
s = input("str: ")
print(s[::-1][::2])
```

```python
#5. Fibonacci series till 1000:
def fibo(a, b):
    res = [a, b]
    while True:
        nxt = res[-1] + res[-2]
        if nxt > 1000:
            break
        res.append(nxt)
    return res
a = int(input("a: "))
b = int(input("b: "))
print(fibo(a, b))
```

```python
#6. filter evens:
def filter_even(lst):
    return [x for x in lst if x % 2 == 0]
print(filter_even([1,2,3,4,5,6]))
```

```python
#7. square with docstring:
def square(n):
    """Return square of n"""
    return n * n
print(square(5))
```

```python
#8. modify global variable:
g = 10
def mod_global():
    global g
    g = 20
mod_global()
print(g)
```

```python
#15. recursive factorial:
def rec_factorial(n):
    if n == 0:
        return 1
    return n * rec_factorial(n - 1)
print(rec_factorial(5))
```

```python
#16. sort tuples by age:
people = [("Alice", 25), ("Bob", 20), ("Carol", 30)]
sorted_people = sorted(people, key=lambda x: x[1])
print(sorted_people)
```

```python
#17. median:
def median(lst):
    lst = sorted(lst)
    n = len(lst)
    if n % 2 == 1:
        return lst[n // 2]
    else:
        return (lst[n // 2 - 1] + lst[n // 2]) / 2
print(median([3, 1, 4, 2, 5]))
```
