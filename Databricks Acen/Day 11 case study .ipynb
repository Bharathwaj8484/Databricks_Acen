```python
#1. Read customer_product.csv file using Pandas
import pandas as pd
df=pd.read_csv("customer_product.csv")
print(df)
```

```python
#2. Print first 10 rows of dataframe
print(df.head(10))
```

```python
#3. Print no. of rows and columns available in the dataset
print(df.shape[0])
print(df.shape[1])
```

```python
#4. Print information of dataset df.info
print(df.info)
```

```python
#5. Print statistical summary of dataset
print(df.describe())
```

```python
# Check data types of each column
print(df.dtypes)
```

```python
#7. Check any column has missing values or not and if yes, then handle them
print(df.isnull())
print(df.dropna())

'''
print(df.fillna(df.mean()))
print(df.fillna(df.median()))
print(df.fillna(df.mode()))
'''
```

```python
'''8.'''

total_amount=df['Purchase_Amount'].sum()
avg_amount=df['Purchase_Amount'].mean()
max_amount=df['Purchase_Amount'].max()
min_amount=df['Purchase_Amount'].min()
std_amount=df['Purchase_Amount'].std()
print("Total purchase amount: ",total_amount)
print("Average purchase amount: ",avg_amount)
print("Maximum purchase amount: ",max_amount)
print(" Minimum purchase amount: ",min_amount)
print("Standard Deviation of purchase amount: ",std_amount)
```

```python
Category_amount=df.groupby('Product_Category')['Purchase_Amount'].sum()
print(Category_amount)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.bar(df['Product_Category'],df['Purchase_Amount'],color='cyan')
plt.ylabel('total purchase amount')
plt.xlabel('Product Category')
plt.title('Total purchase amount for each product category')
plt.show()
```

```python
'''11. Data visualization – Histogram for purchase amount distribution'''
plt.figure(figsize=(8,6))
plt.hist(df['Purchase_Amount'],color='pink',edgecolor='black')
plt.xlabel('Purchase Amount')
plt.ylabel('Quantity')
plt.title("Histogram for purchase amount distribution")
plt.show()
```

```python
'''12. Data visualization – Line chart for purchase trends'''
purchase_trends=df.groupby('Date')['Purchase_Amount'].sum().reset_index()
purchase_trends
plt.figure(figsize=(8,6))
plt.plot(purchase_trends['Date'],purchase_trends['Purchase_Amount'],color='red')
plt.xlabel('Date')
plt.ylabel('Purchase Amount')
plt.title('Line chart for purchase trends')
plt.show()
```

```python
'''13. Data visualization – Pie chart for product categories'''
plt.figure(figsize=(8,6))
colors=['red','blue','pink']
plt.pie(Category_amount.values,labels=Category_amount.index,colors=colors)
plt.title('Pie chart for product categories')
plt.show()
```
