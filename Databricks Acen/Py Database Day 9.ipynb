```python
import mysql.connector
import csv
import os
# -------- Section 1: Student Table (Local DB "testdb") --------
conn1 = mysql.connector.connect(host="localhost", user="root", password="root", database="testdb")
cur1 = conn1.cursor()
```

```python
#1. Create table student
cur1.execute("DROP TABLE IF EXISTS student")
cur1.execute("CREATE TABLE student (studentid INT PRIMARY KEY, studentname VARCHAR(50), studentage INT)")
```

```python
#2. Insert 5 records from user input
for i in range(5):
    sid = int(input("sid: "))
    sname = input("name: ")
    sage = int(input("age: "))
    cur1.execute("INSERT INTO student VALUES (%s, %s, %s)", (sid, sname, sage))
conn1.commit()
```

```python
#3. Print all data
cur1.execute("SELECT * FROM student")
for row in cur1.fetchall():
    print(row)
```

```python
#4. Update record for given studentid
uid = int(input("update sid: "))
newname = input("new name: ")
newage = int(input("new age: "))
cur1.execute("UPDATE student SET studentname=%s, studentage=%s WHERE studentid=%s", (newname, newage, uid))
conn1.commit()
```

```python
#5. Delete record for given studentid
did = int(input("delete sid: "))
cur1.execute("DELETE FROM student WHERE studentid=%s", (did,))
conn1.commit()

cur1.close()
conn1.close()
```

```python
# -------- Section 2: School DB from students.csv --------
conn2 = mysql.connector.connect(host="localhost", user="root", password="root")
cur2 = conn2.cursor()

# Create database school and table students
cur2.execute("DROP DATABASE IF EXISTS school")
cur2.execute("CREATE DATABASE school")
cur2.execute("USE school")
cur2.execute("DROP TABLE IF EXISTS students")
cur2.execute("""CREATE TABLE students (
    student_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    age INT,
    grade INT,
    subject VARCHAR(50)
)""")
```

```python
# Insert data from students.csv (ensure file exists in same folder)
with open("students.csv", "r") as f:
    r = csv.reader(f)
    next(r)  # skip header
    for row in r:
        cur2.execute("INSERT INTO students VALUES (%s, %s, %s, %s, %s, %s)", tuple(row))
conn2.commit()
```

```python
# Fetch and display all records
cur2.execute("SELECT * FROM students")
for row in cur2.fetchall():
    print(row)
```

```python
# Fetch names of students in grade 10
cur2.execute("SELECT first_name, last_name FROM students WHERE grade=10")
for row in cur2.fetchall():
    print(row)
```

```python
# Total number of students
cur2.execute("SELECT COUNT(*) FROM students")
print("Total students:", cur2.fetchone()[0])
```

```python
# Update subject for student_id=4 to Biology
cur2.execute("UPDATE students SET subject='Biology' WHERE student_id=4")
conn2.commit()
```

```python
# Delete record for student_id=3
cur2.execute("DELETE FROM students WHERE student_id=3")
conn2.commit()

cur2.close()
conn2.close()
```

```python
# -------- Section 3: Store DB from products.csv --------
conn3 = mysql.connector.connect(host="localhost", user="root", password="root")
cur3 = conn3.cursor()

cur3.execute("DROP DATABASE IF EXISTS store")
cur3.execute("CREATE DATABASE store")
cur3.execute("USE store")
cur3.execute("DROP TABLE IF EXISTS products")
cur3.execute("""CREATE TABLE products (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(100),
    category VARCHAR(50),
    price DECIMAL(10,2),
    stock INT
)""")
with open("products.csv", "r") as f:
    r = csv.reader(f)
    next(r)
    for row in r:
        cur3.execute("INSERT INTO products VALUES (%s, %s, %s, %s, %s)", tuple(row))
conn3.commit()
```

```python
# Fetch all product details
cur3.execute("SELECT * FROM products")
for row in cur3.fetchall():
    print(row)
```

```python
# Names of products in Electronics
cur3.execute("SELECT product_name FROM products WHERE category='Electronics'")
for row in cur3.fetchall():
    print(row[0])
```

```python
# Total stock value
cur3.execute("SELECT SUM(price * stock) FROM products")
print("Total stock value:", cur3.fetchone()[0])
```

```python
# Increase price of Appliances by 10%
cur3.execute("UPDATE products SET price = price * 1.1 WHERE category='Appliances'")
conn3.commit()
```

```python
# Delete product with product_id=5
cur3.execute("DELETE FROM products WHERE product_id=5")
conn3.commit()

cur3.close()
conn3.close()
```

```python
# -------- Section 4: Travel Agency DB from trips.csv --------
conn4 = mysql.connector.connect(host="localhost", user="root", password="root")
cur4 = conn4.cursor()

cur4.execute("DROP DATABASE IF EXISTS travel_agency")
cur4.execute("CREATE DATABASE travel_agency")
cur4.execute("USE travel_agency")
cur4.execute("DROP TABLE IF EXISTS trips")
cur4.execute("""CREATE TABLE trips (
    trip_id INT PRIMARY KEY,
    origin VARCHAR(50),
    destination VARCHAR(50),
    distance_km INT,
    price DECIMAL(10,2)
)""")
with open("trips.csv", "r") as f:
    r = csv.reader(f)
    next(r)
    for row in r:
        cur4.execute("INSERT INTO trips VALUES (%s, %s, %s, %s, %s)", tuple(row))
conn4.commit()
```

```python
# Fetch all trip details
cur4.execute("SELECT * FROM trips")
for row in cur4.fetchall():
    print(row)
```

```python
# Trips with distance > 400
cur4.execute("SELECT * FROM trips WHERE distance_km > 400")
for row in cur4.fetchall():
    print(row)
```

```python
# Average price of trips
cur4.execute("SELECT AVG(price) FROM trips")
print("Avg price:", cur4.fetchone()[0])
```

```python
# Update price of trips from Dallas to Houston (-10%)
cur4.execute("UPDATE trips SET price = price * 0.9 WHERE origin='Dallas' AND destination='Houston'")
conn4.commit()
```

```python
# Delete trips costing less than 50
cur4.execute("DELETE FROM trips WHERE price < 50")
conn4.commit()

cur4.close()
conn4.close()
```

```python
# -------- Section 5: Bookings & Trips (Travel Agency) --------
conn5 = mysql.connector.connect(host="localhost", user="root", password="root")
cur5 = conn5.cursor()
```

```python
# Bookings table
cur5.execute("DROP TABLE IF EXISTS Bookings")
cur5.execute("""CREATE TABLE Bookings (
    booking_id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    destination VARCHAR(100),
    travel_date DATE,
    price DECIMAL(10,2)
)""")
cur5.execute("INSERT INTO Bookings VALUES (1, 'Alice', 'Paris', '2024-01-15', 1200.00)")
cur5.execute("INSERT INTO Bookings VALUES (2, 'Bob', 'London', '2024-02-10', 900.00)")
cur5.execute("INSERT INTO Bookings VALUES (3, 'Charlie', 'New York', '2024-03-05', 1500.00)")
cur5.execute("INSERT INTO Bookings VALUES (4, 'Diana', 'Tokyo', '2024-04-20', 2000.00)")
cur5.execute("INSERT INTO Bookings VALUES (5, 'Ethan', 'Dubai', '2024-05-12', 1800.00)")
conn5.commit()
```

```python
# Query Bookings where price > 1000 and destination is Paris or New York
cur5.execute("SELECT * FROM Bookings WHERE price > 1000 AND (destination='Paris' OR destination='New York')")
for row in cur5.fetchall():
    print(row)
```

```python
# Trips table with extra fields: trip_id, agency_name, destination, num_customers, total_revenue
cur5.execute("DROP TABLE IF EXISTS Trips")
cur5.execute("""CREATE TABLE Trips (
    trip_id INT PRIMARY KEY,
    agency_name VARCHAR(100),
    destination VARCHAR(100),
    num_customers INT,
    total_revenue DECIMAL(10,2)
)""")
cur5.execute("INSERT INTO Trips VALUES (1, 'TravelPro', 'Paris', 50, 60000.00)")
cur5.execute("INSERT INTO Trips VALUES (2, 'Explorer Inc', 'London', 30, 45000.00)")
cur5.execute("INSERT INTO Trips VALUES (3, 'Wanderlust', 'New York', 40, 72000.00)")
cur5.execute("INSERT INTO Trips VALUES (4, 'TravelPro', 'Tokyo', 20, 40000.00)")
cur5.execute("INSERT INTO Trips VALUES (5, 'Explorer Inc', 'Dubai', 25, 50000.00)")
conn5.commit()
```

```python
# Total revenue by agency
cur5.execute("SELECT agency_name, SUM(total_revenue) FROM Trips GROUP BY agency_name")
for row in cur5.fetchall():
    print(row)
```

```python
# Most popular destination (highest num_customers)
cur5.execute("SELECT destination, num_customers FROM Trips ORDER BY num_customers DESC LIMIT 1")
print("Popular:", cur5.fetchone())
```

```python
# Join Bookings and Trips on destination
cur5.execute("""SELECT b.customer_name, b.destination, b.travel_date, t.agency_name, t.total_revenue
FROM Bookings b JOIN Trips t ON b.destination = t.destination""")
for row in cur5.fetchall():
    print(row)

cur5.close()
conn5.close()
```
