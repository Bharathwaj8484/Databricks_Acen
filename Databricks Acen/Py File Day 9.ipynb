```python
#3. data to CSV:
data_lines = []
while True:
    inp = input("data (or stop): ")
    if inp.lower() == "stop":
        break
    data_lines.append(inp)
with open("data.csv", "w") as f:
    for d in data_lines:
        f.write(d + "\n")
```

```python
#4. word file copy:
from docx import Document
doc = Document("sample.docx")
for para in doc.paragraphs:
    print(para.text)
doc_copy = Document()
for para in doc.paragraphs:
    doc_copy.add_paragraph(para.text)
doc_copy.save("copy.docx")
```

```python
#10. dict to CSV:
import csv
d = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Frank", "Grace", "Hannah", "Ivy", "Jack"],
    "Age": [25, 30, 22, 28, 35, 40, 32, 29, 26, 38],
    "City": ["Paris", "London", "New York", "Tokyo", "Dubai", "Los Angeles", "Berlin", "Sydney", "Barcelona", "Rome"]
}
with open("user_data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Age", "City"])
    for i in range(len(d["Name"])):
        writer.writerow([d["Name"][i], d["Age"][i], d["City"][i]])
```

```python
#11. read CSV:
with open("user_data.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
#14. split data to files:
data_list = [
    "Alice, 25, Paris",
    "Bob, 30, London",
    "Charlie, 22, New York",
    "Diana, 28, Tokyo",
    "Ethan, 35, Dubai",
    "Frank, 40, Los Angeles",
    "Grace, 32, Berlin",
    "Hannah, 29, Sydney",
    "Ivy, 26, Barcelona",
    "Jack, 38, Rome"
]
mid = len(data_list) // 2
group1 = data_list[:mid]
group2 = data_list[mid:]
with open("group1.txt", "w") as f:
    for line in group1:
        f.write(line + "\n")
with open("group2.txt", "w") as f:
    for line in group2:
        f.write(line + "\n")
```

```python
#15. find & replace:
with open("city_data.txt", "r") as f:
    lines = f.readlines()
new_lines = [line.replace("New York", "San Francisco") for line in lines]
with open("city_data.txt", "w") as f:
    f.writelines(new_lines)
```
